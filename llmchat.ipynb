```python
!pip install llama-cpp-python

import os
from llama_cpp import Llama

# Path to the model file within the Kaggle dataset
dataset_path = "/kaggle/input/minithinky-v2-1bllama-3-2-q8-0-gguf"
model_file = "MiniThinky-v2-1B-Llama-3.2-Q8_0.gguf" # The filename *inside* the dataset
model_path = os.path.join(dataset_path, model_file) # full path

# Check if the model file exists within the dataset
if not os.path.exists(model_path):
    print(f"Error: Model file not found at {model_path}. Ensure you've correctly loaded your dataset and the filename matches!")
else:
   llm = Llama(
        model_path=model_path,
        n_gpu_layers=16  # Offload all layers to GPU
    )

   while True:
        user_prompt = input("User: ")
        if user_prompt.lower() in ["exit", "quit", "bye"]:
             break

        output = llm(
            prompt=user_prompt,
            max_tokens=150,
            stop=["User:","\n"],
            echo=False
        )

        print("Model:", output["choices"][0]["text"].strip())

   print ("Exiting")
```
